## 0. Setup e Carregamento

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 100

CORES_RISCO = {"baixo": "#2ecc71", "medio": "#f39c12", "alto": "#e74c3c"}

df = pd.read_parquet("../data/dataset_safefield.parquet")
print(f"Carregado: {df.shape[0]} registros x {df.shape[1]} colunas")
```

```python
print("=== NULOS ===")
null_df = pd.DataFrame({"nulos": df.isnull().sum(), "%": (df.isnull().sum() / len(df) * 100).round(1)})
print(null_df[null_df["nulos"] > 0])

n_eq = df["equipamento_id"].nunique()
print(f"
=== EQUIPAMENTOS ===")
print(f"Unicos: {n_eq}  |  Media avaliacoes/equip: {len(df)/n_eq:.1f}")

print(f"
=== PERIODO ===")
print(f"Inicio: {df[chr(39)]timestamp[chr(39)].min().date()}  Fim: {df[chr(39)]timestamp[chr(39)].max().date()}")
```
